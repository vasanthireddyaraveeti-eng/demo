```python
med = df["Age"].median()
df["Age"] = df["Age"].fillna(med)
df.head(25)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Job Title'] = le.fit_transform(df['Job Title'])
print(df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
x = df[['Job Title', 'Years of Experience']]
y = df.Salary
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
mlr = LinearRegression()
mlr.fit(x_train,y_train)
```

```python
y_pred=mlr.predict(x_test)
print(y_pred)
```

```python
print("R2 Score:", r2_score(y_test,y_pred))
```

## Project Report: Salary Prediction using Linear Regression

### 1. Project Title
Salary Prediction Model using Machine Learning

### 2. Objective
The primary objective of this project is to develop a machine learning model capable of predicting an individual's salary based on key professional attributes such as their Job Title and Years of Experience. This model aims to provide insights into salary determinants and offer a predictive tool for estimating potential earnings.

### 3. Dataset Description
The project utilizes a dataset loaded from `/content/archive (3).zip`. The dataset contains information about individuals, including:
*   **Age**: The age of the individual (float).
*   **Gender**: The gender of the individual (object).
*   **Education Level**: The educational attainment of the individual (object).
*   **Job Title**: The professional role of the individual (object).
*   **Years of Experience**: The total years of professional experience (float).
*   **Salary**: The salary of the individual (float), which is the target variable for prediction.

The dataset initially contains 375 entries and 6 columns. After preprocessing, 373 valid entries were used for model training.

### 4. Data Preprocessing Steps Performed
The following steps were taken to clean and prepare the data:

*   **Loading Data**: The dataset was loaded into a pandas DataFrame from a CSV file within a zip archive.
*   **Missing Value Identification**: `df.isna().sum()` was used to identify columns with missing values. It was found that 'Age', 'Gender', 'Education Level', 'Job Title', 'Years of Experience', and 'Salary' each had 2 missing values.
*   **Handling Missing Values (Age)**: The missing values in the 'Age' column were imputed using the median age of the dataset (36.0).
*   **Handling Missing Values (Remaining Columns)**: All rows containing any remaining `NaN` values across other columns were dropped using `df.dropna()`. This ensured that the model was trained on complete records.
*   **Feature Encoding ('Job Title')**: The 'Job Title' column, being a categorical feature, was transformed into numerical format using `LabelEncoder` from `sklearn.preprocessing`. Each unique job title was assigned a unique integer value.

### 5. Feature Selection Process
For this linear regression model, the following features were selected from the preprocessed dataset as independent variables (X):

*   `Job Title` (after Label Encoding)
*   `Years of Experience`

The `Salary` column was chosen as the dependent variable (Y), which the model aims to predict.

### 6. Machine Learning Algorithms Used

*   **Linear Regression**: A simple yet effective supervised machine learning algorithm, `LinearRegression` from `sklearn.linear_model`, was chosen for this predictive task. It models the relationship between a dependent variable and one or more independent variables by fitting a linear equation to the observed data.

### 7. Model Training and Testing Procedure

1.  **Data Splitting**: The preprocessed data was split into training and testing sets using `train_test_split` from `sklearn.model_selection`. A test size of 20% was used, with `random_state=42` for reproducibility.
    *   `x_train`, `y_train`: Used to train the model (298 samples).
    *   `x_test`, `y_test`: Used to evaluate the model's performance on unseen data (75 samples).
2.  **Model Initialization**: An instance of `LinearRegression` was created.
3.  **Model Training**: The linear regression model was trained using the `x_train` and `y_train` datasets (`mlr.fit(x_train, y_train)`).
4.  **Prediction**: After training, the model made predictions on the test set (`y_pred = mlr.predict(x_test)`).

### 8. Accuracy of Each Model

*   **Linear Regression**: The model achieved an R2 Score of **0.8975** (approximately 89.75%).

### 9. Best Performing Model and Justification

Since only one model, Linear Regression, was employed in this project, it is considered the best performing model by default. The R2 score of approximately 0.8975 indicates that the model explains about 89.75% of the variance in salary, which is a strong performance for a simple linear model. This suggests that 'Job Title' and 'Years of Experience' are significant predictors of salary within this dataset.
